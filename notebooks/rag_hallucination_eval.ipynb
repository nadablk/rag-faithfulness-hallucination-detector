# RAG Faithfulness & Hallucination Evaluation
This notebook demonstrates claim-level NLI evaluation and pandas reporting.

```python
from src.detector import RAGFaithfulnessDetector
import pandas as pd
```

```python
detector = RAGFaithfulnessDetector()
query = "When was the Eiffel Tower completed?"
contexts = [
    "The Eiffel Tower was completed in 1889 for the Exposition Universelle."
]
answer = "The Eiffel Tower was completed in 1889. It was designed by Gustave Eiffel."

result = detector.evaluate(query, contexts, answer)
print("Faithfulness score:", result["faithfulness_score"])
display(detector.to_dataframe(result))
```

## Example report
The claim-level DataFrame contains entailment, contradiction, neutral, support status, and entity discrepancy signals.

```python
report = detector.to_dataframe(result)
summary = pd.DataFrame([{
    "query": result["query"],
    "faithfulness_score": result["faithfulness_score"],
    "supported_claims": result["supported_claims"],
    "total_claims": result["total_claims"],
}])
display(summary)
```
